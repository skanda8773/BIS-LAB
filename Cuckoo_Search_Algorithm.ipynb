```python

import random
import networkx as nx

G = nx.DiGraph()
edges = [(0,1,1),(1,3,2),(1,2,1),(2,4,2),
         (3,4,1),(4,5,2),(4,6,2),(5,7,2),
         (6,7,1),(7,9,2),(8,10,2),(9,10,1)]

G.add_weighted_edges_from(edges)

paths = list(nx.all_simple_paths(G, 0, 10))

def fitness(p):
    return sum(G[u][v]['weight'] for u,v in zip(p,p[1:]))

n = 10
nests = [random.choice(paths) for _ in range(n)]

for i in range(100):
    j = random.randrange(n)
    step = random.expovariate(1)
    new = random.choice(paths[:]) if step > 0 else nests[j]

    if fitness(new) < fitness(nests[j]):
        nests[j] = new

    nests.sort(key=fitness)

    for j in range(2):
        nests[-(j+1)] = random.choice(paths)

best = min(nests, key=fitness)

print("Best Path:", best)
print("Minimum Cost:", fitness(best))
```
