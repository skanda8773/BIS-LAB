```python

import random, math

points = [(0,0), (2,3), (5,4), (6,1), (3,-2), (-2,2)]
n, vehicles, capacity = 5, 2, 3

def dist(a,b):
    return math.dist(points[a], points[b])

pher = [[1]*6 for _ in range(6)]
best, best_cost = None, float('inf')

for _ in range(100):
    for ant in range(20):
        cities = list(range(1,n+1))
        random.shuffle(cities)
        routes = [cities[i:i+capacity] for i in range(0,n,capacity)]
        cost = 0

        for r in routes:
            path = [0] + r + [0]
            cost += sum(dist(path[i],path[i+1])
                        for i in range(len(path)-1))

        if cost < best_cost:
            best, best_cost = routes, cost

        for r in routes:
            path = [0] + r + [0]
            for a,b in zip(path,path[1:]):
                pher[a][b] += 1/cost

    for i in range(6):
        for j in range(6):
            pher[i][j] *= 0.5

print("Best Delivery Routes:")
for i,r in enumerate(best):
    print(f"Vehicle {i+1}: Warehouse ->",
          " -> ".join(map(str,r)), "-> Warehouse")

print("Minimum Distance:", round(best_cost,2))
```
