```python
#grey wolf optimization(GWO)

import numpy as np

def gwo():
    n, dim, iterations = 10, 2, 100
    wolves = np.random.uniform(-5, 5, (n, dim))

    for t in range(iterations):
        fitness = np.sum(wolves ** 2, axis=1)
        order = np.argsort(fitness)

        alpha = wolves[order[0]].copy()
        beta = wolves[order[1]].copy()
        delta = wolves[order[2]].copy()

        a = 2 - 2 * t / iterations

        for i in range(n):
            positions = []
            for leader in [alpha, beta, delta]:
                r1, r2 = np.random.rand(), np.random.rand()
                A = 2 * a * r1 - a
                C = 2 * r2
                D = abs(C * leader - wolves[i])
                positions.append(leader - A * D)

            wolves[i] = np.clip(
                np.mean(positions, axis=0), -5, 5
            )

    best = wolves[np.argmin(np.sum(wolves ** 2, axis=1))]
    print("Best solution:", best)
    print("Minimum fitness:", np.sum(best ** 2))

gwo()
```
