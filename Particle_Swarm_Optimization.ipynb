```python
#Particle Swarm Optimization for Function Optimization

import numpy as np
def objective_function(x):
    return x**2
NUM_PARTICLES = 10
ITERATIONS = 50
W = 0.7
C1 = 1.5
C2 = 1.5
LOWER_BOUND = -10
UPPER_BOUND = 10
positions = np.random.uniform(
    LOWER_BOUND, UPPER_BOUND, NUM_PARTICLES
)
velocities = np.random.uniform(
    -1, 1, NUM_PARTICLES
)
personal_best_positions = positions.copy()
personal_best_values = objective_function(positions)
best_index = np.argmin(personal_best_values)
global_best_position = personal_best_positions[best_index]
global_best_value = personal_best_values[best_index]
for iteration in range(ITERATIONS):
    r1 = np.random.random(NUM_PARTICLES)
    r2 = np.random.random(NUM_PARTICLES)
    velocities = (
        W * velocities
        + C1 * r1 * (personal_best_positions - positions)
        + C2 * r2 * (global_best_position - positions)
    )
    positions = positions + velocities
    positions = np.clip(
        positions, LOWER_BOUND, UPPER_BOUND
    )
    fitness = objective_function(positions)
    improved = fitness < personal_best_values
    personal_best_positions[improved] = positions[improved]
    personal_best_values[improved] = fitness[improved]
    best_index = np.argmin(personal_best_values)
    if personal_best_values[best_index] < global_best_value:
        global_best_position = personal_best_positions[best_index]
        global_best_value = personal_best_values[best_index]
    print(
        f"Iteration {iteration + 1:02d} | "
        f"Best x: {global_best_position:.4f} | "
        f"Best Fitness: {global_best_value:.4f}"
    )
print("\n" + "=" * 40)
print("Best Solution Found:")
print(f"Best x       : {global_best_position:.4f}")
print(f"Minimum Value: {global_best_value:.4f}")
print("=" * 40)
```
